```python
import sys
from importlib.metadata import version

assert sys.prefix != sys.base_prefix, "Selectează kernelul din .venv"

print("Interpreter:", sys.executable)

for package in ("duckdb", "plotly", "ipykernel", "nbformat"):
    print(f"{package}: {version(package)}")
```

```python
import duckdb

con = duckdb.connect(
    config={
        "memory_limit": "2GB",
        "threads": 2,
    }
)

result = con.execute("SELECT 40 + 2").fetchone()

assert result == (42,)
print("DuckDB funcționează:", result)
```

```python
import plotly.graph_objects as go

fig = go.Figure(
    go.Bar(
        x=["A", "B", "C"],
        y=[10, 20, 15],
    )
)

fig.update_layout(
    title="Verificare setup — date demonstrative",
    xaxis_title="Categorie",
    yaxis_title="Valoare",
)

fig.show(renderer="plotly_mimetype")
```

```python
from pathlib import Path

cwd = Path.cwd().resolve()
project_root = cwd if (cwd / "requirements.in").is_file() else cwd.parent

assert (project_root / "requirements.in").is_file(), (
    "Deschide în VS Code folderul rădăcină al proiectului."
)

output = project_root / "results" / "verificare_setup.html"

fig.write_html(output, include_plotlyjs=True, full_html=True)

assert output.is_file()
print("Grafic exportat:", output)
```
